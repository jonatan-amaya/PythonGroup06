```python
#1. Create and display a dictionary using the data from the aforementioned statement

population = {
    "Lima": 10092000,
    "Arequipa": 1008000,
    "Trujillo": 919000
}

print(population)
```

```python
#2. Only display the dictionary's keys

print(population.keys())
```

```python
#3. Get and display the population of Lima

print(population.get("Lima"))
```

```python
#4. Add Cusco and its population

population.update({"Cusco": 428000})

print(population)
```

```python
#5. Remove "Trujillo"

population.pop("Trujillo")
```

```python
#6. Print the final version of the dictionary

print(population)
```
