```python
# Print the original list

grades = [15, 18, 12, 20, 16]

print("The grades are:", grades)
```

```python
# Add the grade 17

grades.append(17)
```

```python
# Sort from lowest to highest

grades.sort(reverse=False)
```

```python
# Display:

# the highest grade;
print("The highest grade is:", max(grades))

# the lowest grade;
print("The lowest grade is:", min(grades))

# the total number of grades.
print("The total number of grades is:", len(grades))
```

```python
# Print the final list

print("The updated grades are:", grades)
```
