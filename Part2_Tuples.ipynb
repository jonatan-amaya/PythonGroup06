```python
# 1. Print the tuple
ages = (20, 25, 22, 30, 28)

print("Tuple:", ages)
```

```python
# 2. Display the first and last elements of the tuple
print("First element:", ages[0])
print("Last element:", ages[-1])
```

```python
# 3. Display:
# the maximum value;
# the minimum value;
# the number of elements.
print("Maximum value:", max(ages))
print("Minimum value:", min(ages))
print("Number of elements:", len(ages))
```

```python
# 4. Try to modify one of the elements of the tuple.
ages[0] = 21
```
